```python
from pathlib import Path
import hashlib
from collections import defaultdict

dataset_path = Path("../data/images/raw")

valid_extensions = [".jpg", ".jpeg", ".png", ".webp"]

hash_to_files = defaultdict(list)

# Step 1: Calculate fingerprint for every image
for folder in dataset_path.iterdir():

    if not folder.is_dir():
        continue

    for image_path in folder.iterdir():

        if image_path.suffix.lower() not in valid_extensions:
            continue

        with open(image_path, "rb") as f:
            file_hash = hashlib.md5(f.read()).hexdigest()

        hash_to_files[file_hash].append(image_path)


# Step 2: Find duplicate groups
duplicate_groups = []

for file_hash, files in hash_to_files.items():

    if len(files) > 1:
        duplicate_groups.append(files)


print("Duplicate groups:", len(duplicate_groups))

total_duplicate_files = sum(
    len(group) - 1 for group in duplicate_groups
)

print("Duplicate files to potentially remove:", total_duplicate_files)
```

```python
cross_class_duplicates = []

for group in duplicate_groups:

    classes = set()

    for image_path in group:
        classes.add(image_path.parent.name)

    if len(classes) > 1:
        cross_class_duplicates.append(group)


print("Cross-class duplicate groups:", len(cross_class_duplicates))
```

```python
class_counts = {}

for folder in dataset_path.iterdir():

    if not folder.is_dir():
        continue

    count = 0

    for image_path in folder.iterdir():

        if image_path.suffix.lower() in valid_extensions:
            count += 1

    class_counts[folder.name] = count


print("Image count by class:\n")

for class_name, count in class_counts.items():
    print(f"{class_name}: {count}")
```

```python
from PIL import Image

bad_images = []

for folder in dataset_path.iterdir():

    if not folder.is_dir():
        continue

    for image_path in folder.iterdir():

        if image_path.suffix.lower() not in valid_extensions:
            continue

        try:
            with Image.open(image_path) as img:
                img.load()

        except Exception as e:
            bad_images.append((image_path, str(e)))


print("Images that cannot be opened:", len(bad_images))

for image_path, error in bad_images:
    print(image_path)
    print("Error:", error)
```

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
import random

dataset_path = Path("../data/images/raw")

valid_extensions = [".jpg", ".jpeg", ".png", ".webp"]

random.seed(42)

for folder in sorted(dataset_path.iterdir()):

    if not folder.is_dir():
        continue

    images = [
        p for p in folder.iterdir()
        if p.suffix.lower() in valid_extensions
    ]

    selected_images = random.sample(
        images,
        min(5, len(images))
    )

    fig, axes = plt.subplots(1, len(selected_images), figsize=(15, 4))

    if len(selected_images) == 1:
        axes = [axes]

    for ax, image_path in zip(axes, selected_images):

        img = Image.open(image_path)

        ax.imshow(img)
        ax.set_title(image_path.name)
        ax.axis("off")

    fig.suptitle(
        f"Class: {folder.name}",
        fontsize=16
    )

    plt.tight_layout()
    plt.show()
```
