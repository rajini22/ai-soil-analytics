```python
# Create the final structured soil-test dataset

structured_data = soil_raw[
    [
        "NUMBER",
        "CROP",
        "Moisture (%dw)",
        "pH (Ext 1:2.5)",
        "organic Matter (%)",
        "total-N (%)",
        "avail-P (mg/kg)",
        "avail-K (mg/kg)"
    ]
].copy()

structured_data.columns = [
    "sample_id",
    "crop",
    "moisture",
    "ph",
    "organic_matter",
    "nitrogen",
    "phosphorus",
    "potassium"
]

print("Rows:", structured_data.shape[0])
print("Columns:", structured_data.shape[1])

structured_data.head()
```

```python
# Remove records missing any required soil parameter

before = len(structured_data)

structured_data = structured_data.dropna(
    subset=[
        "moisture",
        "ph",
        "organic_matter",
        "nitrogen",
        "phosphorus",
        "potassium"
    ]
).copy()

after = len(structured_data)

print("Records before cleaning:", before)
print("Records after cleaning:", after)
print("Records removed:", before - after)

print("\nMissing values:")
print(structured_data.isna().sum())
```
