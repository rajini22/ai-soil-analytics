```python
# Check for outliers using the IQR method

soil_features = [
    "moisture",
    "ph",
    "organic_matter",
    "nitrogen",
    "phosphorus",
    "potassium"
]

print("Outlier check")
print("-" * 40)

for column in soil_features:
    Q1 = structured_data[column].quantile(0.25)
    Q3 = structured_data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = (
        (structured_data[column] < lower) |
        (structured_data[column] > upper)
    ).sum()

    print(f"{column}: {outliers} outliers")
```
